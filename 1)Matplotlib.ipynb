```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**Matplotlib Basics**

```python
x = np.linspace(0,5,11)
x
y = x**2
y
```

```python
plt.plot(x,y)
plt.title('hello')
plt.xlabel('number')
plt.ylabel('square')
```

**Subplots in Matplotlib**

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x**3,y)
plt.subplot(2,2,4)
plt.plot(x**2,x**2)
```

**Object Oriented way**

```python
fig = plt.figure()
axes1 = fig.add_axes([0.1,0.1,0.8,0.7])
axes1.plot(x,y)
axes2 = fig.add_axes([0.2,0.48,0.3,0.3])
axes2.plot(x,y)
```

**Types of plot**

```python
plt.scatter(x,y)
```

```python
plt.hist(y)
```

```python
plt.boxplot(x)
```

```python
fig = plt.figure()
ax = fig.add_axes([0.1,0.1,0.8,0.8])
ax.plot(x,y)
plt.savefig("basic.png")
```

**Working with Images**

```python
import matplotlib.image as mpimg
```

```python
img = mpimg.imread('basic.png')
plt.imshow(img)
plt.axis('off')
```
