```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = sns.load_dataset('tips')
df
```

```python
sns.countplot(x = df['sex'], hue = df['smoker'])
```

```python
sns.barplot(x= df['sex'], y = df['total_bill'])
```

```python
sns.barplot(x= df['sex'], y = df['total_bill'], estimator= np.sum)
```

```python
sns.barplot(x= df['sex'], y = df['tip'], estimator= np.sum)
```

```python
sns.boxplot(data= df, x = 'tip', hue = 'day', palette='rainbow')
```

```python
sns.violinplot(data=df, y = 'total_bill', x = 'day', palette='rainbow')
```
